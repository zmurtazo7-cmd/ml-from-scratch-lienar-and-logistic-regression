```python
df.columns = df.columns.str.strip().str.replace('"', '')

x = df['study hours'].values
y = df['scores'].values
```

```python
w = 0
b = 0
lr = 0.01
```

```python
np.random.seed(42)
idx = np.random.permutation(len(x))
split = int(0.8 * len(x))
train_idx, test_idx = idx[:split], idx[split:]

x_train, x_test = x[train_idx], x[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
```

```python
w = 0
b = 0
lr = 0.01
```

```python
for step in range(1000):
    y_pred = w * x_train + b

    loss = np.mean((y_pred - y_train) ** 2)

    dw = np.mean(2 * (y_pred - y_train) * x_train)
    db = np.mean(2 * (y_pred - y_train))

    w = w - lr * dw
    b = b - lr * db

    if step % 100 == 0:
        print(f"Step {step}, b: {b:.4f} Loss: {loss:.4f}")

print(f"Final w: {w:.4f}, Final b: {b:.4f}")
```

```python
y_test_pred = w * x_test + b

test_mse = np.mean((y_test_pred - y_test) ** 2)
print(f"Test MSE: {test_mse:.4f}")

for i in range(10):
    print(f"Predicted: {y_test_pred[i]:.2f}, Actual: {y_test[i]:.2f}")
```
